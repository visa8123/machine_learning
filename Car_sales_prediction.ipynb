```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv('car_purchasing.csv', encoding='latin1')
#drop the column
df = df.drop(columns=['customer name', 'customer e-mail'])

#clean the data
df = pd.get_dummies(df, columns=['country'], drop_first=True)

X = df.drop('car purchase amount', axis=1)
y = df['car purchase amount']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Train model
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
rmse = mean_squared_error(y_test, y_pred, squared=False)
r2 = r2_score(y_test, y_pred)

print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

RMSE ~ 2385.60: This is low relative to target (car purchase amount), indicating the model makes small errors on average.
R² Score ~ 0.95: The model explains ~95% of the variance in car purchase amounts.
